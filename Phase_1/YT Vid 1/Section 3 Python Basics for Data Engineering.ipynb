```python
# lambda functions 
# this is used when we want to write small piece of function
add = lambda x,y: x + y # add two numbers
print(add(3,5))
```

```python
# using map functionality in functions like lambda 
# map maps the function to the complete iterable

#example 1
numbers = [1, 2, 3, 4] # the iterable
squared = list(map(lambda x : x**2, numbers))  # squares each number
print(squared)

#example 2

words = ['hello', 'world']
up_words = list(map(lambda x: x.upper(), words))  # converts each word to its uppercase
print(up_words)
```
