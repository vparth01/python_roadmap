```python
# Text Files handling 

# writing
with open('datasets/sample_text.txt', 'w') as file:
    file.write('Hello, mere pyaare saaptiyon!\n')
    file.write('Is baar kar ke hi rahenge.\n')

# reading
with open('datasets/sample_text.txt', 'r') as file:
    content = file.read()
    print('Content of the text file:\n', content)
```

```python
# CSV Files handling 

import pandas as pd
# writing
product_data = pd.DataFrame(
    {
        'Product': ['Widget A', 'Widget B', 'Widget'],
        'Price': [10.99, 14.99, 7.99],
        'Stock': [50, 20, 100]
    }
)
product_data.to_csv('datasets/product_data.csv', index=False)

# reading
df = pd.read_csv('datasets/product_data.csv')
print("contents of the product_data.csv:\n", df)
```
